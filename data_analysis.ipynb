```python
import clickhouse_connect
import polars as pl

client = clickhouse_connect.get_client(host="localhost", port=8123, username="default", password="changeme")
```

```python
result = client.query("SELECT * FROM fact_subscription_events where platform = 'fenster'")
pl.DataFrame(result.result_rows, schema=result.column_names, orient="row")
```

```python
result = client.query("SELECT * FROM dim_customer where platform = 'apfel'")
pl.DataFrame(result.result_rows, schema=result.column_names, orient="row")
```
