```python
import clickhouse_connect
import polars as pl

client = clickhouse_connect.get_client(host="localhost", port=8123, username="default", password="changeme")
```

```python
monthly_metrics = client.query("""
    SELECT 
        *
    FROM monthly_metrics
    """
)
monthly_metrics_df = pl.DataFrame(monthly_metrics.result_rows, schema=monthly_metrics.column_names, orient="row")

monthly_metrics_df.filter((pl.col("platform") == "apfel") & (pl.col('subscription_type') == 'premium'))
```

```python
# Clean up Clickhouse
client.query("DROP TABLE IF EXISTS fact_subscription_events")
client.query("DROP TABLE IF EXISTS dim_user")
client.query("DROP TABLE IF EXISTS monthly_metrics")
```
